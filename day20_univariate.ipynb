```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

# Categorical Data

### A. Countplot

```python
sns.countplot(x='Embarked', hue='Embarked', data=df, palette=['red', 'green', 'orange'], legend=False)
# df['Embarked'].value_counts().plot(kind='bar')
```

### Pie Chart

```python
df['Embarked'].value_counts().plot(kind='pie', autopct='%.2f')
```

# Numerical Data

### A. Histogram

```python
# You can create ranges and then you can display them as categorical data

import matplotlib.pyplot as plt

plt.hist(df['Fare'], bins=50)
```

### B. Displot

```python
# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Probability Density Function (PDF)
# This shows the distribution of values (similar to histogram with kde)
sns.histplot(df['Age'], kde=True, stat="density", ax=ax1)
ax1.set_title('Probability Density Function (PDF) of Age')
ax1.set_xlabel('Age')
ax1.set_ylabel('Density')

# Plot 2: Cumulative Distribution Function (CDF)
# This shows the cumulative probability up to each value
sns.ecdfplot(df['Age'], ax=ax2)
ax2.set_title('Cumulative Distribution Function (CDF) of Age')
ax2.set_xlabel('Age')
ax2.set_ylabel('Cumulative Probability')

plt.tight_layout()
plt.show()
```

### C. Boxplot

```python
sns.boxplot(x=df['Age'], orient='h')
```
