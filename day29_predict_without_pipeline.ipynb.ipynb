```python
import pickle as pkl
import numpy as np
```

```python
ohe_sex = pkl.load(open('day29_models/ohe_sex.pkl', 'rb'))
ohe_embarked = pkl.load(open('day29_models/ohe_embarked.pkl', 'rb'))
clf = pkl.load(open('day29_models/clf.pkl', 'rb'))
```

```python
test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'], dtype='object').reshape(1,7)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
print(f"ohe_sex features_in: {ohe_sex.n_features_in_}")
print(f"ohe_embarked features_in: {ohe_embarked.n_features_in_}")
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_transformed = np.concatenate((
    test_input[:, [0, 3, 4, 5]], 
    test_input_age, 
    test_input_sex, 
    test_input_embarked
), axis=1)
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
