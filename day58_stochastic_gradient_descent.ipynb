```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
x, y = load_diabetes(return_X_y=True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=2
)
```

```python
linear_regression = LinearRegression()
```

```python
linear_regression.fit(x_train, y_train)
```

```python
y_pred = linear_regression.predict(x_test)
```

```python
print(r2_score(y_test, y_pred))
```

```python
print(linear_regression.coef_)
```

```python
print(linear_regression.intercept_)
```

# Batch Gradient Descent

```python
class BatchGradientDescent:
    def __init__(self, learning_rate=0.01, epochs=50):
        self.coef_ = None
        self.intercept_ = None
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, x_train, y_train):

        # For gradient descent we need to run a loop of epochs times
        # In each iteration need to find intercept and coefs
        # To find intercept 
        # Intercept_new = Intercept_old  - learning_rate * derivative_of_intercept_old
        # derivative_of_intercept_old = -2 / n sum of error
        # To find coefs
        # derivative_of_coef = -2 / n * sum of error * x

        n_samples, n_features = x_train.shape
        print(n_samples, n_features)

        self.intercept_ = 0
        self.coef_ = np.zeros(n_features)
        
        for i in range(self.epochs):
            y_pred = np.dot(x_train, self.coef_) + self.intercept_
            error = y_train - y_pred

            der_intercept = (-2 / n_samples) * np.sum(error)
            self.intercept_ -= self.learning_rate * der_intercept

            der_coefs = (-2 / n_samples) * np.dot(error, x_train)
            self.coef_ -= self.learning_rate * der_coefs

        print(f"Intercept: {self.intercept_}")
        print(f"Coefs: {self.coef_}")

    def predict(self, x_test):
        return np.dot(x_test, self.coef_) + self.intercept_
            
```

```python
batch_gd = BatchGradientDescent(learning_rate=0.1, epochs=1000)
```

```python
batch_gd.fit(x_train, y_train)
```

```python
y_pred_bgd = batch_gd.predict(x_test)
```

```python
print(r2_score(y_test, y_pred_bgd))
```

# Stochastic Gradient Descent

```python
class StochasticGradientDescent:
    def __init__(self, learning_rate=0.01, epochs=50):
        self.coef_ = None
        self.intercept_ = None
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, x_train, y_train):

        # For gradient descent we need to run a loop of epochs times
        # In each iteration need to find intercept and coefs
        # To find intercept 
        # Intercept_new = Intercept_old  - learning_rate * derivative_of_intercept_old
        # derivative_of_intercept_old = -2 / n sum of error
        # To find coefs
        # derivative_of_coef = -2 / n * sum of error * x

        n_samples, n_features = x_train.shape
        print(n_samples, n_features)

        self.intercept_ = 0
        self.coef_ = np.zeros(n_features)
        
        for i in range(self.epochs):
            for j in range(n_samples):
                random_id = np.random.randint(0, n_samples)
                
                y_pred = np.dot(x_train[random_id], self.coef_) + self.intercept_
                error = y_train[random_id] - y_pred

                der_intercept = -2 * error
                self.intercept_ -= self.learning_rate * der_intercept

                der_coefs = -2 * error * x_train[random_id]
                self.coef_ -= self.learning_rate * der_coefs

        print(f"Intercept: {self.intercept_}")
        print(f"Coefs: {self.coef_}")

    def predict(self, x_test):
        return np.dot(x_test, self.coef_) + self.intercept_
            
```

```python
stochastic_gd = StochasticGradientDescent(learning_rate=0.1, epochs=100)
```

```python
stochastic_gd.fit(x_train, y_train)
```

```python
y_pred_sgd = stochastic_gd.predict(x_test)
```

```python
print(r2_score(y_test, y_pred_sgd))
```
